```python
import random

randomnum = random.randint(1,100)
print ("I have a number from 1-100. You have 7 attempts.")
attempts = 0

while True:

  try:
    if attempts == 7:
      print (f"You ran out of attempts, the number was {randomnum}")

    userguess = input(f"Guess {attempts + 1}:")

    guess = int(userguess)
    attempts += 1

    if guess == randomnum:
      print (f"You got it correct! Attempts: {attempts}")
      break
    elif guess > randomnum:
      print ("Too high!")
    elif guess < randomnum:
      print ("Too low!")
  except ValueError:
    print ("Thats not a number, so it doesn't count as an attempt.")

```
