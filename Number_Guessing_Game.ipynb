```python
import random

randomnum = random.randint(1,100)
print ("I have a number from 1-100. Try to get it within 7 attempts.")
attempts = 0

while True:

  try:

    if attempts == 7:
      print (f"You ran out of attempts, the number was {randomnum}")
      tryagain = input('Would you like to try again? Say "Yes" or "No".')

      if tryagain.lower() == "yes":
        print ("I have a number from 1-100. Try to guess it within 7 attempts!")
        attempts = 0
      elif tryagain.lower() == "no":
        print ("Ok! I will be ending the game now.")
        break

    userguess = input(f"Guess {attempts + 1}:")
    guess = int(userguess)
    attempts += 1

    if guess == randomnum:
      print (f"You got it correct! Attempts: {attempts}")
      continue
    elif guess > randomnum:
      print ("Too high!")
    elif guess < randomnum:
      print ("Too low!")

  except ValueError:
    print ("Thats not a number, so it doesn't count as an attempt.")

```
