```python
def square_complex(a, b):
  new_a = a * a - b * b
  new_b = 2 * a * b


  return new_a, new_b

print(square_complex(2,1))



# c = 2 + i
# number is first one plus second one square root of all that
```

```python
def generate_mandelbrot(a, b):
  ## find or list the first 20 numbers
  z = [0,0]
  for i in range (20):
    z = square_complex(z[0], z[1])
    z[0] += a
    z[1] += b
    print(z)
```
