```python
food = int(input("How much food do you have (0-5):"))
water = int(input("How much water do you have (0-10):"))
fitness = int(input("How fit are you (0-10):"))
weapon = input("Do you have a weapon (0-15):")

weaponscore = 0

if weapon.lower() = "yes":
  weaponscore += 15
elif weapon.lower() = "no"
  continue
else:
  "I was unable to complete the score because your answer for the weapon was incorrect."


officialscore = ((food + water + fitness + weaponscore) + "/40")

print (officialscore)
```
