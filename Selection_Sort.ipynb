```python
list1 = [4, 1, 9, 8, 8, 5, 4, 18, 3, 19, 14, 2, 1, 3, 7, 8, 17, 20, 1, 18, 7, 18, 14, 8, 15, 19, 9, 1, 6, 14]


minimumindex = 0
minimum = list1[0]


for i in (list1):
  for j in range(len(list1) - minimumindex):
    if list1[j + minimumindex] < minimum:
      minimum = list1[j + minimumindex]
  minimumindex += 1

  list1[minimumindex-1], minimum = minimum, list1[minimumindex-1]


print(list1)
```
