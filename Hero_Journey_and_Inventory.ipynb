```python
player ={
    "name" : "Adventureboy",
    "health" : 100,
    "weaponscore" : 1,
    "location" :  "Area 1"
}

inventory = ["Block of Wood"]

def showstatus():
  print ("\n === PLAYERSTATUS ===")
  print (f"Name: {player['name']}")
  print (f"Health: {player['health']}")
  print (f"Weapon Level: {player['weaponscore']}")

def intro():
  print ("This is a Riddle Survivor game. It will consist of riddles where you ahve to solve riddles in order to get out of the forest.")
  print ("\n You are a traveler alone in the forest, and you have 10 days to survive before the army comes to save you.")
  print ("You will need to avoid predators, kill animals for food, build survival materials using things you find, and stay safe in shelter.")

def riddle1():

  option1 = "a map"
  option2 = "a globe"
  option3 = "a mirage"
  answer = option1

  question = input(f"\n\nI have cities, but no houses. I have mountains, but no trees. I have water, but no fish. What am I?: {option1}, {option2}, {option3}. \n")
  if question.lower() == answer:
    print ("Correct")
    player ['location'] = "Area 2"
    print (f"You are moving on to {player['location']}")
  elif question == option1 or option2 or option3:
    print ("Wrong answer. You fail.")
  else:
    print ("That is not a valid option")



intro()
showstatus()
riddle1()
```
